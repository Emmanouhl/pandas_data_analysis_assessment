*PART B - Investigate Data Problems*

```python
# Count missing values column by column
missing_patient_id = df["Patient_ID"].isnull().sum()
print(f"Number of missing Patient ID: {missing_patient_id}")

missing_patient_name = df["Patient_Name"].isnull().sum()
print(f"Number of missing Patient Name: {missing_patient_name}")

missing_age = df["Age"].isnull().sum()
print(f"Number of missing Age: {missing_age}")

missing_gender = df["Gender"].isnull().sum()
print(f"Number of missing Gender: {missing_gender}")

missing_city = df["City"].isnull().sum()
print(f"Number of missing City: {missing_city}")

missing_visit_count = df["Visit_Count"].isnull().sum()
print(f"Number of missing Visit Counts: {missing_visit_count}")

missing_record_year = df["Record_Year"].isnull().sum()
print(f"Number of missing Record Years: {missing_record_year}")

missing_visit_type = df["Visit_Type"].isnull().sum()
print(f"Number of missing Visit Type: {missing_visit_type}")
```

*PART C - Data Cleaning*
=

```python

# Create a copy so we don't modify the original raw data (good practice)
df_cleaned = df.copy()

print("Starting cleaning process...\n")

# ==========================================
# STEP 1: Handle Potentially Unnecessary Records
# ==========================================

# A. Remove Test/Dummy records based on Patient ID (9000+ IDs are clearly test data)
df_cleaned = df_cleaned[~df_cleaned['Patient_ID'].between(9000, 9999)]

# B. Remove rows where Patient_Name indicates a test
df_cleaned = df_cleaned[~df_cleaned['Patient_Name'].str.contains('Test|Dummy|System', case=False, na=False)]

# C. Remove exact duplicate rows
df_cleaned = df_cleaned.drop_duplicates()

# ==========================================
# STEP 2: Clean Missing Values
# ==========================================

# A. Fill missing 'City' values with "Unknown" instead of deleting the patient record
df_cleaned['City'] = df_cleaned['City'].fillna('Unknown')

# B. Drop rows where critical numerical data is missing (Age or Visit_Count)
# You cannot accurately analyze a patient without their age or visit count
df_cleaned = df_cleaned.dropna(subset=['Age', 'Visit_Count'])

# ==========================================
# STEP 3: Clean Unexpected Text & Formatting
# ==========================================

# A. Standardize City names (Capitalize first letter, make the rest lowercase)
# This fixes 'lagos' -> 'Lagos' and 'ABUJA' -> 'Abuja'
df_cleaned['City'] = df_cleaned['City'].str.title()

# B. Fix 'Visit_Count' text values before converting to numbers
# Replace words with their numerical equivalents
visit_count_map = {
    'six': '6',
    'seven': '7',
    'ten': '10',
    'four': '4',
    'N/A': None,  # Treat N/A as missing so we can handle it
    '4 visits': '4'
}

# Apply the mapping
df_cleaned['Visit_Count'] = df_cleaned['Visit_Count'].replace(visit_count_map)

# Now convert the entire column to numeric. 
# Any remaining text that we missed will become NaN (missing).
df_cleaned['Visit_Count'] = pd.to_numeric(df_cleaned['Visit_Count'], errors='coerce')

# Since we converted 'N/A' to NaN, drop those rows now
df_cleaned = df_cleaned.dropna(subset=['Visit_Count'])

# ==========================================
# STEP 4: Clean Numerical Columns
# ==========================================

# A. Fix the Age column
# The test record had Age 0, which is invalid for a patient dataset. 
# We already removed test records, but let's ensure no other 0 ages exist.
df_cleaned = df_cleaned[df_cleaned['Age'] > 0]

# B. Ensure Patient_ID and Record_Year are integers (they were floats like 1745.0)
df_cleaned['Patient_ID'] = df_cleaned['Patient_ID'].astype(int)
df_cleaned['Age'] = df_cleaned['Age'].astype(int)
df_cleaned['Visit_Count'] = df_cleaned['Visit_Count'].astype(int)
df_cleaned['Record_Year'] = df_cleaned['Record_Year'].astype(int)

# ==========================================
# FINAL CHECK
# ==========================================
print("Cleaning complete!")
print(f"Original rows: {len(df)}")
print(f"Cleaned rows: {len(df_cleaned)}")
print(f"Rows removed: {len(df) - len(df_cleaned)}")

# Verify there are no more missing values in critical columns
print("\nRemaining missing values in critical columns:")
print(df_cleaned[['Patient_ID', 'Age', 'City', 'Visit_Count', 'Record_Year']].isnull().sum())

# Display the first few rows of the cleaned data
print("\n--- Cleaned Data Preview ---")
print(df_cleaned.head())
```

*PART D - Patient Classification*
=

```python
# Flag patients with 5+ visits as "Frequent Patient", everyone else as "Occasional Patient"
df_cleaned["Patient_Category"] = np.where(
    df_cleaned["Visit_Count"] >= 5,
    "Frequent Patient",
    "Occasional Patient"
)
df_cleaned["Patient_Category"].value_counts()
```

*PART E - Data Analysis*
=

```python
# Number of (cleaned) patient records per city
df_cleaned["City"].value_counts()
```

```python
# Filter to frequent patients only and count them
frequent_patients = df_cleaned[df_cleaned["Patient_Category"] == "Frequent Patient"]
print(f"Total frequent patients: {len(frequent_patients)}")
```

```python
# Average number of visits across all cleaned patient records
average_visits = df_cleaned["Visit_Count"].mean()
print(f"Average visit count: {average_visits}")
```

```python
# .isnull().any(axis=1) flags a row True if ANY column in that row is missing,
# so this counts how many rows (in the original, uncleaned df) had at least one missing value
records_with_missing = df.isnull().any(axis = 1).sum()
print(f"Total records with missing info: {records_with_missing}")
```

```python
# Number of rows left after cleaning
remaining_records = len(df_cleaned)
print(f"Total records remaining: {remaining_records}")
```

```python
# Group by patient category and compute, for each group:
# - the count of patients (via Patient_ID)
# - the average number of visits
summary_table = df_cleaned.groupby("Patient_Category").agg(
    Number_of_Patients = ("Patient_ID", "count"),
    Average_Visits = ("Visit_Count", "mean")
).reset_index()
print(summary_table)
```

*Summary of the table:*
=

The summary table shows that the hospital has a significantly larger number of Frequent Patients compared to Occasional Patients, and these frequent patients average nearly 8 visits each. This indicates a strong base of loyal, returning patients who likely require ongoing or chronic care. The hospital should prioritize resources and staffing toward managing these frequent visitors to ensure high-quality continuity of care

*PART A - Load and Inspect Data*
=

```python
# Load the banking transaction dataset
df = pd.read_csv("Banking_Customer_Transation.csv")
```

```python
# Preview the first 5 rows
df.head()
```

```python
# (rows, columns) of the dataset
df.shape
```

```python
# Column names, non-null counts and dtypes
df.info()
```

```python
# Summary statistics for numeric columns
df.describe()
```

*PART B - Investige the Transaction Amount*
=

```python
# Show only the columns that have at least one missing value, with their missing counts
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])
```

```python
print(f"\n--- PART B: INVESTIGATION ---")

# 1. Look for Missing Values
missing_amounts = df["Amount"].isnull().sum()
print(f"1. Missing values in 'Amount': {missing_amounts}")

# 2. Look for Unexpected Text
# We try to convert the column to numeric. Anything that fails (becomes NaN) is unexpected text.
# We also check for common text placeholders like 'N/A' or 'Error'
unexpected_text_mask = pd.to_numeric(df["Amount"], errors = "coerce").isnull() & df["Amount"].notnull()
unexpected_text_df = df[unexpected_text_mask]
print(f"\n2. Unexpected text found in {len(unexpected_text_df)} rows. Unique values:")
print(unexpected_text_df["Amount"].unique())

# 3. Look for Invalid Records
# Invalid records include negative amounts, zero amounts, or extreme outliers that look like system errors.
# First, let's create a temporary numeric column to check
temp_numeric = pd.to_numeric(df["Amount"], errors = "coerce")

# Check for negative numbers (refunds might be valid, but we need to flag them)
negative_amounts = temp_numeric < 0
print(f"\n3a. Negative amounts found: {negative_amounts.sum()}")

# Check for zero amounts (usually invalid for a completed transaction)
zero_amounts = temp_numeric == 0
print(f"3b. Zero amounts found: {zero_amounts.sum()}")

# Check for extreme outliers (e.g., anything over 1 million, which is suspiciously high based on the data)
extreme_outliers = temp_numeric > 1000000
print(f"3c. Extreme outliers (> 1M) found: {extreme_outliers.sum()}")
```

*PART C - Data Cleaning*
=

```python
print("\n--- PART C: CLEANING ---")

# Create a copy to keep the original data intact
df_cleaned = df.copy()

# 1. Clean Missing Values and Unexpected Text
# We will replace known text errors with NaN, then drop rows where Amount is missing.
# Why? We cannot calculate averages or transaction categories without a valid amount.
text_errors = ['Error', 'N/A', 'Pending', 'Not Available', 'Unknown', '##VALUE!', '1O000', 'â_x0082_¦12,500']
df_cleaned["Amount"] = df_cleaned["Amount"].replace(text_errors, np.nan)

# Convert the column to numeric. 
df_cleaned["Amount"] = pd.to_numeric(df_cleaned["Amount"], errors = "coerce")

# Drop rows where Amount is now missing (originally missing or was text)
df_cleaned = df_cleaned.dropna(subset = ["Amount"])

# 2. Handle Invalid Records (Negatives and Zeroes)
# Why? A negative transaction amount usually indicates a refund or reversal, which is 
# conceptually different from a standard transaction. For this analysis, we will remove them.
# Zero amounts are invalid because a completed transaction must have a value.
df_cleaned = df_cleaned[df_cleaned["Amount"] > 0]

# 3. Handle Extreme Outliers
# Why? Transactions over 1,000,000 are extremely rare and likely due to data entry errors 
# or system glitches (like the 672,537.71 value). Removing them prevents skewed averages.
df_cleaned = df_cleaned[df_cleaned["Amount"] <= 1000000]

print(f"Cleaned dataset shape: {df_cleaned.shape}")
print(f"Rows removed: {df.shape[0] - df_cleaned.shape[0]}")
```

*PART D -Transaction Classification*
=

```python
print(f"\n--- PART D: CLASSIFICATION ---")

# Create the 'Transaction_Category' column
# Condition: Amount >= 100,000 -> 'Large Transaction'
# Condition: Amount < 100,000 -> 'Regular Transaction'
df_cleaned["Transaction_Category"] = np.where(
    df_cleaned["Amount"] >= 100000, 
    "Large Transaction", 
    "Regular Transaction"
)

# Display the results
print(f"\nTransaction Category Counts:")
print(df_cleaned["Transaction_Category"].value_counts())

print(f"\nPreview of Cleaned and Classified Data:")
print(df_cleaned[["Customer ID", "Amount", "Transaction_Category"]].head())
```

*PART E - Business Analysis*
=

```python
# ==========================================
# 1. How many transactions occurred in each city?
# ==========================================
print(f"1. Number of Transactions by City:")
transactions_by_city = df_cleaned["City"].value_counts()
print(transactions_by_city)
print("-" * 40)

# ==========================================
# 2. What is the average transaction amount?
# ==========================================
print(f"\n2. Average Transaction Amount:")
avg_transaction = df_cleaned["Amount"].mean()
print(f"${avg_transaction:,.2f}")
print("-" * 40)

# ==========================================
# 3. How many large transactions occurred?
# ==========================================
print(f"\n3. Number of Large Transactions:")
# We can count using the category we created earlier, or by filtering the amount
large_transactions_count = df_cleaned[df_cleaned["Transaction_Category"] == "Large Transaction"]
print(f"{len(large_transactions_count)} large transactions occurred.")
print("-" * 40)

# ==========================================
# 4. Which transaction type appears most frequently?
# ==========================================
print(f"\n4. Most Frequent Transaction Type:")
most_frequent_type = df_cleaned["Transaction Type"].mode()[0]
print(f"'{most_frequent_type}' appears most frequently.")
print("-" * 40)

# ==========================================
# 5. What is the average transaction amount by transaction type?
# ==========================================
print(f"\n5. Average Transaction Amount by Transaction Type:")
avg_by_type = df_cleaned.groupby("Transaction Type")["Amount"].mean().sort_values(ascending=False)
# Format the output to look like currency
print(avg_by_type.apply(lambda x: f"${x:,.2f}"))
print("-" * 40)
```

*Challenge Task*
=

```python
# 1. Use groupby() to determine the average transaction amount by city
avg_amount_by_city = df_cleaned.groupby("City")["Amount"].mean().sort_values(ascending=False)

# 2. Identify the city with the highest average transaction amount
highest_avg_city = avg_amount_by_city.index[0]
highest_avg_value = avg_amount_by_city.iloc[0]

# Display the results
print(f"--- Average Transaction Amount by City ---")
print(avg_amount_by_city.apply(lambda x: f"${x:,.2f}"))

print(f"\n--- Conclusion ---")
print(f"The city with the highest average transaction amount is: {highest_avg_city} (${highest_avg_value:,.2f})")
```

*What this means from a business-analysis perspective:*
=

The finding that Benin City has the highest average transaction amount suggests that while there may be a high volume of transactions in other cities, the customers in Benin City are conducting significantly higher-value transactions. From a business perspective, this indicates that Benin City is likely the primary hub for High-Net-Worth Individuals (HNWIs) or large corporate accounts.

This insight is crucial for strategic decision-making: the bank should prioritize retaining and upselling premium services (like wealth management or tailored business accounts) in Benin City to maximize revenue. Conversely, if the bank wants to increase revenue in other cities, it may need to investigate why transaction values are lower there—perhaps due to a lack of premium products or a customer base that is more retail-focused.

*Energy Consumption Analysis (Energy / Utilities)*
=

*PART A - Load and Inspect DataSet*
=

```python
# 1. Load the dataset
df = pd.read_csv("Energy_Consumption.csv")

print("--- PART A: UNDERSTAND THE DATA ---")

# 2. Number of records
print(f"Number of records: {df.shape[0]}")

# 3. Number of columns
print(f"Number of columns: {df.shape[1]}")

# 4. Data types
print("\nData Types:")
print(df.dtypes)

# 5. Summary statistics
print("\nSummary Statistics:")
# We only get statistics for numeric columns. Notice 'Monthly Consumption' is an object due to text errors.
print(df.describe(include='all'))
```

*PART B - Data Cleaning*
=

```python
print(f"\n--- PART B: DATA CLEANING ---")

# Create a copy to keep the original data intact
df_cleaned = df.copy()

# 1. Investigate & Handle Missing City Values
print(f"1. Missing City values: {df_cleaned['City'].isnull().sum()}")
# Decision: Drop rows where City is missing. City is crucial for regional analysis.
df_cleaned = df_cleaned.dropna(subset=['City'])

# 2. Investigate & Handle Missing Consumption Values
# Check for blank spaces, 'N/A', 'Unknown', 'No Reading', 'Error', 'Pending', '##VALUE!', '12O', '5,2O0'
# We will replace these with NaN and drop them, as we cannot analyze consumption without data.
missing_placeholders = ['N/A', 'Unknown', 'No Reading', 'Error', 'Pending', '##VALUE!', '12O', '5,2O0', ' ']
df_cleaned['Monthly Consumption'] = df_cleaned['Monthly Consumption'].replace(missing_placeholders, np.nan)

# Convert the column to numeric (this will turn remaining text into NaN)
df_cleaned['Monthly Consumption'] = pd.to_numeric(df_cleaned['Monthly Consumption'], errors='coerce')

# Drop rows where Consumption is now NaN
df_cleaned = df_cleaned.dropna(subset=['Monthly Consumption'])

# 3. Handle Unnecessary Rows
# Drop duplicates (if any)
df_cleaned = df_cleaned.drop_duplicates()

# Drop negative consumption values (e.g., -45.0) and zero values, as they are invalid for consumption
df_cleaned = df_cleaned[df_cleaned['Monthly Consumption'] > 0]

print(f"Data cleaned. Remaining records: {len(df_cleaned)}")
```

*PART C - Consumption Category*
=

```python
print(f"\n--- PART C: CREATE CONSUMPTION CATEGORY ---")

# Create the 'Consumption_Category' column using np.where
df_cleaned['Consumption_Category'] = np.where(
    df_cleaned['Monthly Consumption'] >= 500, 
    'High Consumption', 
    'Normal Consumption'
)

print("Category Counts:")
print(df_cleaned['Consumption_Category'].value_counts())
```

*PART D - Analyze Consumption*
=

```python
print(f"\n--- PART D: ANALYZE CONSUMPTION ---")

# 1. Number of customers in each city
print(f"1. Number of Customers by City:")
print(df_cleaned['City'].value_counts())
print("-" * 40)

# 2. Average consumption
avg_consumption = df_cleaned['Monthly Consumption'].mean()
print(f"2. Average Consumption: {avg_consumption:.2f} kWh")
print("-" * 40)

# 3. Number of high-consumption customers
high_consumption_count = df_cleaned[df_cleaned['Consumption_Category'] == 'High Consumption'].shape[0]
print(f"3. Number of High-Consumption Customers: {high_consumption_count}")
print("-" * 40)

# 4. Average consumption by customer type
print(f"4. Average Consumption by Customer Type:")
avg_by_type = df_cleaned.groupby('Customer Type')['Monthly Consumption'].mean().sort_values(ascending=False)
print(avg_by_type.apply(lambda x: f"{x:.2f} kWh"))
print("-" * 40)

# 5. Number of customers using each energy source
print(f"5. Number of Customers by Energy Source:")
print(df_cleaned['Energy Source'].value_counts())
```

*Challenge Task*
=

```python
# 1. Create the summary table using groupby and agg
summary_table = df_cleaned.groupby('Customer Type').agg(
    Average_Consumption=('Monthly Consumption', 'mean'),
    Number_of_Customers=('Customer ID', 'count')
).reset_index()

# 2. Rename columns to match the challenge prompt
summary_table.columns = ['Customer Type', 'Average Consumption', 'Number of Customers']

# 3. Format the Average Consumption to 2 decimal places for readability
summary_table['Average Consumption'] = summary_table['Average Consumption'].round(2)

# 4. Sort by Average Consumption (highest to lowest) to make the pattern obvious
summary_table = summary_table.sort_values(by='Average Consumption', ascending = False)

# Display the final summary table
print(f"--- Consumption Summary by Customer Type ---")
print(summary_table.to_string(index = False))
```

*Recommendation:*
=
The analysis clearly shows that Industrial and Commercial customers have significantly higher average consumption rates compared to Residential customers, despite being a much smaller portion of the total customer base.

The energy company should investigate the load patterns and peak usage times of these high-consumption Industrial customers. Specifically, they need to determine if these industries are placing excessive strain on the local grid during peak hours, which could lead to outages or require expensive infrastructure upgrades. Investigating this pattern could lead to implementing "Time-of-Use" tariffs for industrial clients to encourage them to shift their heavy machinery operations to off-peak hours, thereby stabilizing the grid and improving overall operational efficiency.
